```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age

s1 = Student("Arpit", 19)

print(s1.name)
print(s1.age)
```

## __repr__ Method
Provides developer-friendly representation.

```python
class Student:

    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student('{self.name}')"

s = Student("Arpit")

print(repr(s))
```

## Instance Methods
Methods that work on a particular object.

```python
class Person:

    def __init__(self, name):
        self.name = name

    def greet(self):
        print(f"Hello, {self.name}")

p = Person("Arpit")

p.greet()
```

## Class Variables
Shared among all objects.

```python
class Student:

    college = "Chandigarh University"

    def __init__(self, name):
        self.name = name

s1 = Student("Arpit")
s2 = Student("Rohan")

print(s1.college)
print(s2.college)
```

## Class Methods
Methods that work on the class itself.

```python
class Student:

    college = "CU"

    @classmethod
    def change_college(cls, new_name):
        cls.college = new_name

Student.change_college("IIT Delhi")

print(Student.college)
```

## Inheritance
A child class inherits properties and methods from a parent class.

```python
class Animal:

    def sound(self):
        print("Animal sound")

class Dog(Animal):

    def bark(self):
        print("Woof")

d = Dog()

d.sound()
d.bark()
```

## Method Overriding
Child class can redefine parent methods.

```python
class Animal:

    def sound(self):
        print("Animal sound")

class Dog(Animal):

    def sound(self):
        print("Woof")

d = Dog()

d.sound()
```
