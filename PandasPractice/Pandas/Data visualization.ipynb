```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
%matplotlib inline
```

```python
df = pd.read_csv("../Datasets/cleaned_titanic.csv")
```

```python
df.head()
```

```python
df["Age"].plot(kind="hist", bins=20)
plt.show()
```

```python
df["Survived"].value_counts().plot(kind="bar")
plt.show()
```

```python
x=[1,2,3,4,5]
y=[10,5,7,8,13]

plt.plot(x,y)
plt.title("Simple Line Chart")
plt.xlabel("X axis")
plt.ylabel("Y axis")
plt.show()
```

```python
corr= df.corr(numeric_only=True)

sns.heatmap(corr,annot=True)

plt.show()
```

```python
sns.boxplot(df["Age"])

plt.show()
```

```python
plt.scatter(df["Age"], df["Fare"])

plt.xlabel("Age")
plt.ylabel("Fare")
plt.show()
```

Extracted Insights
1. Most passengers were between 20–40 years old.
2. Higher fare passengers had better survival chances.
3. Pclass negatively correlates with Fare.
4. Many outliers exist in Fare column.
5. Survival count is lower than death count.
