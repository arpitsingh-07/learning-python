```python
import pandas as pd
```

```python
students = pd.DataFrame({
    "ID": [1, 2, 3],
    "Name": ["Arpit", "Rahul", "Subdil"]
})

students
```

```python
marks = pd.DataFrame({
    "ID": [1, 2, 3],
    "marks": ["90", "75", "87"]
})

marks
```

```python
pd.merge(students, marks, on="ID")
```

```python
pd.merge(students, marks, on="ID", how="inner")
```

```python
pd.merge(students, marks, on="ID", how="right")
```

```python
pd.merge(students, marks, on="ID", how="outer")
```

```python
df1 = pd.DataFrame({
    "Name": ["Arpit", "Rahul"]
})

df2 = pd.DataFrame({
    "Name": ["Priya", "Aman"]
})

pd.concat([df1, df2])
```

```python
pd.concat([df1, df2], axis=1)
```

```python
players = pd.DataFrame({
    "PlayerID": [1, 2, 3, 4],
    "Name": ["Mbappe", "Messi", "Ronaldo", "Haaland"],
    "Club": ["Real Madrid", "Inter Miami", "Al Nassr", "Man City"]
})
```

```python
stats = pd.DataFrame({
    "PlayerID": [1, 2, 3, 5],
    "Goals": [30, 25, 20, 35],
    "Assists": [10, 15, 8, 5]
})
```

```python
print(pd.merge(players, stats, on="PlayerID", how="inner"))
```

```python
print(pd.merge(players, stats, on="PlayerID", how="left"))
```

```python
print(pd.merge(players, stats, on="PlayerID", how="outer"))
```

```python
extra = pd.DataFrame({
    "PlayerID": [6],
    "Name": ["Vinicius"],
    "Club": ["Real Madrid"]
})
print(pd.concat([players, extra], ignore_index=True))
```
