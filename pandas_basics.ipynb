```python
data={
    "Name": ["Mbappe","Ronaldo", "Messi", "Lamine Yamal", "Erling Haaland"],
    "Club": ["Real Madrid", "Al Nassr", "Inter Miami", "Barcelona", "Manchester City"],
    "Position": ["LW", "ST", "CAM", "RW", "ST"]
}
df=pd.DataFrame(data)
print(df)
```

```python
print(df["Club"])
```

```python
df["Rating"]=[93,94,94,92,91]
```

```python
print(df[df["Club"]=="Real Madrid"])
```

```python
print(df)
```

```python
print(df.sort_values(by="Rating", ascending=False))
```

```python
print(df[(df["Rating"] > 92) & (df["Position"] == "ST")])
```

```python
print(df[df["Rating"]>92][["Name", "Club"]])
```
