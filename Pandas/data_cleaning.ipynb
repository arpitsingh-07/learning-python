```python
import pandas as pd
```

```python
df = pd.read_csv("Titanic-Dataset.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
df["Age"]=df["Age"].fillna(df["Age"].mean())
```

```python
df=df.drop(columns=["Cabin"])
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Age"].mode()[0])
```

```python
df.isnull().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.info()
```

```python
df.to_csv("cleaned_titanic.csv", index=False)
```
