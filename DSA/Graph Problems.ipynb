# LeetCode 733 - Flood Fill

Problem:

Starting from a given pixel,
change all connected pixels of the same color.

Example:

1 1 1
1 1 0
1 0 1

Start = (1,1)
New Color = 2

Result:

2 2 2
2 2 0
2 0 1

Idea:

1. Store original color
2. Visit connected cells
3. Change color
4. Continue DFS/BFS

```python
def flood_fill(image, sr, sc, new_color):

    old = image[sr][sc]

    if old == new_color:
        return image

    def dfs(r, c):

        if not (0 <= r < len(image) and 0 <= c < len(image[0])):
            return

        if image[r][c] != old:
            return

        image[r][c] = new_color

        dfs(r + 1, c)
        dfs(r - 1, c)
        dfs(r, c + 1)
        dfs(r, c - 1)

    dfs(sr, sc)

    return image
```

```python
image = [
    [1,1,1],
    [1,1,0],
    [1,0,1]
]

print(flood_fill(image, 1, 1, 2))
```

# Flood Fill Logic

Step 1:
Find starting color

old = image[sr][sc]

Step 2:
Visit all connected cells

Step 3:
Change color

image[r][c] = new_color

Step 4:
Move in 4 directions

Up
Down
Left
Right

This is DFS on a grid.

# LeetCode 200 - Number of Islands

Problem:

Count islands in a grid.

1 = Land
0 = Water

Example:

1 1 0 0
1 0 0 1
0 0 1 1

Answer = 2 islands

Idea:

Whenever you find a new land cell:

1. Increase count
2. DFS/BFS entire island
3. Mark visited

```python
def num_islands(grid):

    rows = len(grid)
    cols = len(grid[0])

    count = 0

    def dfs(r, c):

        if not (0 <= r < rows and 0 <= c < cols):
            return

        if grid[r][c] != "1":
            return

        grid[r][c] = "0"

        dfs(r + 1, c)
        dfs(r - 1, c)
        dfs(r, c + 1)
        dfs(r, c - 1)

    for r in range(rows):
        for c in range(cols):

            if grid[r][c] == "1":

                count += 1
                dfs(r, c)

    return count
```

```python
grid = [
    ["1","1","0","0"],
    ["1","0","0","1"],
    ["0","0","1","1"]
]

print(num_islands(grid))
```

# Number of Islands Logic

Loop through every cell.

When you find:

grid[r][c] == "1"

You discovered a NEW island.

Increase count.

Then DFS the entire island and mark it visited.

This prevents counting the same island twice.

# Connected Components

A connected component is a group of nodes where
every node can be reached from another node.

Example:

0 -- 1

2 -- 3

4

Components:

{0,1}
{2,3}
{4}

Answer = 3 components

```python
graph = {
    0:[1],
    1:[0],
    2:[3],
    3:[2],
    4:[]
}

visited = set()

def dfs(node):

    if node in visited:
        return

    visited.add(node)

    for neighbor in graph[node]:
        dfs(neighbor)

components = 0

for node in graph:

    if node not in visited:

        components += 1
        dfs(node)

print("Connected Components =", components)
```

# BFS vs DFS

BFS

- Queue
- Level by level
- Shortest path
- More memory

DFS

- Stack / Recursion
- Goes deep first
- Less memory
- Great for islands and traversal
