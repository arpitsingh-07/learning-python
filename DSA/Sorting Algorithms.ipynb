# Sorting Algorithms

## Topics Covered
- Bubble Sort
- Insertion Sort
- Merge Sort
- Quick Sort
- Time Complexity Comparison

Sorting is the process of arranging elements in a particular order (ascending or descending).

Example:

Input:
[5, 2, 8, 1, 3]

Sorted:
[1, 2, 3, 5, 8]

```python
def bubble_sort(arr):
    n = len(arr)

    for i in range(n):
        swapped = False

        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True

        if not swapped:
            break

    return arr


nums = [64, 34, 25, 12, 22, 11, 90]
print(bubble_sort(nums))
```

## Bubble Sort

Repeatedly compares adjacent elements and swaps them if they are in the wrong order.

Example:

5 3 2

Pass 1:
3 5 2

Pass 2:
3 2 5

Pass 3:
2 3 5

Time Complexity:
- Best: O(n)
- Average: O(n²)
- Worst: O(n²)

```python
def insertion_sort(arr):

    for i in range(1, len(arr)):

        key = arr[i]
        j = i - 1

        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key

    return arr


nums = [12, 11, 13, 5, 6]
print(insertion_sort(nums))
```

## Insertion Sort

Builds a sorted portion one element at a time.

Similar to arranging playing cards in your hand.

Time Complexity:
- Best: O(n)
- Average: O(n²)
- Worst: O(n²)

Better than Bubble Sort for small datasets.

```python
def merge_sort(arr):

    if len(arr) <= 1:
        return arr

    mid = len(arr) // 2

    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    result = []

    i = j = 0

    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    result.extend(left[i:])
    result.extend(right[j:])

    return result


nums = [38, 27, 43, 3, 9, 82, 10]
print(merge_sort(nums))
```

## Merge Sort

Uses Divide and Conquer.

Steps:
1. Divide array into halves.
2. Sort each half recursively.
3. Merge sorted halves.

Time Complexity:
- Best: O(n log n)
- Average: O(n log n)
- Worst: O(n log n)

Very efficient for large datasets.

```python
def quick_sort(arr):

    if len(arr) <= 1:
        return arr

    pivot = arr[len(arr) // 2]

    left = [x for x in arr if x < pivot]
    middle = [x for x in arr if x == pivot]
    right = [x for x in arr if x > pivot]

    return quick_sort(left) + middle + quick_sort(right)


nums = [10, 7, 8, 9, 1, 5]
print(quick_sort(nums))
```

## Quick Sort

Uses a Pivot.

Elements:
- Smaller than pivot → Left
- Greater than pivot → Right

Then recursively sort both sides.

Time Complexity:
- Best: O(n log n)
- Average: O(n log n)
- Worst: O(n²)

In practice, Quick Sort is often one of the fastest sorting algorithms.

```python
sorting_complexities = {
    "Bubble Sort": "O(n²)",
    "Insertion Sort": "O(n²)",
    "Merge Sort": "O(n log n)",
    "Quick Sort (Average)": "O(n log n)"
}

for algo, complexity in sorting_complexities.items():
    print(f"{algo}: {complexity}")
```

# Key Takeaways

Bubble Sort:
- Easy to understand
- Inefficient

Insertion Sort:
- Good for small arrays

Merge Sort:
- Guaranteed O(n log n)

Quick Sort:
- Usually fastest in practice

Rule for Interviews:

Small data:
- Insertion Sort

Large data:
- Merge Sort / Quick Sort

Remember:

O(n log n) is much better than O(n²)
for large inputs.
