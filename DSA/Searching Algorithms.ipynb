Searching Algorithms

Linear search, binary search, and two LeetCode applications.

## Linear search — O(n)
Walk the array, check each element. Works on any array, sorted or not.

```python
def linear_search(arr, target):
    for i, x in enumerate(arr):
        if x == target:
            return i
    return -1
```

## Binary search — O(log n)
Requires a **sorted** array. Halve the search space each step.
- `lo + (hi - lo) // 2` avoids int overflow (matters in Java/C++)
- `while lo <= hi` so the last single element still gets checked
- `mid ± 1` excludes the element we just checked

```python
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = lo + (hi - lo) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1
```

```python
#LeetCode 704
class Solution:
    def search(self, nums, target):
        lo = 0
        hi = len(nums) - 1
        while lo <= hi:
            mid = lo + (hi - lo) // 2

            if nums[mid] == target:
                return mid

            elif nums[mid] < target:
                lo = mid + 1

            else:
                hi = mid - 1

        return -1
```

```python
# LeetCode 35 
class Solution:
    def searchInsert(self, nums, target):
        lo = 0
        hi = len(nums) - 1

        # while lo <= hi so the last single element still gets checked
        while lo <= hi:

            # lo + (hi - lo) // 2 avoids int overflow (matters in Java/C++)
            mid = lo + (hi - lo) // 2

            if nums[mid] == target:
                return mid

            elif nums[mid] < target:
                # mid + 1 excludes the element we just checked
                lo = mid + 1

            else:
                # mid - 1 excludes the element we just checked
                hi = mid - 1

        return lo
```

```python
def searchInsert(nums, target):
    lo, hi = 0, len(nums) - 1
    while lo <= hi:
        mid = lo + (hi - lo) // 2
        if nums[mid] == target:
            return mid
        elif nums[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return lo
```

```python
# quick sanity checks
assert linear_search([5, 2, 9, 1], 9) == 2
assert binary_search([1, 3, 5, 6], 5) == 2
assert binary_search([1, 3, 5, 6], 4) == -1
assert searchInsert([1, 3, 5, 6], 2) == 1
assert searchInsert([1, 3, 5, 6], 7) == 4
print("all passed ✅")
```
