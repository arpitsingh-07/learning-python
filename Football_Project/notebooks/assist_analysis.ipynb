```python
import pandas as pd

df = pd.read_csv("../../Datasets/Football/Player Assists.csv")
```

```python
df.isnull().sum()
```

```python
top_players = df.sort_values(by="Assists", ascending=False)

top_players.head(10)
```

```python
import matplotlib.pyplot as plt

top10 = df.sort_values(by="Assists", ascending=False).head(10)

plt.figure(figsize=(10,5))

plt.bar(top10["Player Name"], top10["Assists"])

plt.title("Top 10 Players by Assists")
plt.xlabel("Player Name")
plt.ylabel("Assists")

plt.xticks(rotation=45)

plt.show()
```

OBSERVATIONS: 
1. Multiple players are tied closely in assists.
2. Messi remains among elite creators.
3. Some players achieved high assists in fewer games.

```python
df["Assist Per Game"] = df["Assists"] / df["Games Played"]

df.head()
```

```python
efficient = df.sort_values(by="Assist Per Game", ascending=False)

efficient.head(10)
```

```python
top_eff = efficient.head(10)

plt.figure(figsize=(10,5))

plt.bar(top_eff["Player Name"], top_eff["Assist Per Game"])

plt.title("Top Players by Assist Per Game")
plt.xlabel("Player")
plt.ylabel("Assist Per Game")

plt.xticks(rotation=45)

plt.show()
```
